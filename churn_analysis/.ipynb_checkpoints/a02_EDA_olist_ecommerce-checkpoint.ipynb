### Lets define churn

#### Churn in the context of this dataset, is customer stopping purchase for a long time

#### customer_unique_id is the id corresponding to each customer and customer_id is an id to associate each transaction to the customer. so one customer can have multiple customer_id but each customer will have an unqiue customer_unique_id. so to find the churned customers, lets join orders and customers on customer_id. 

```python
orders_join = orders.merge(customers[['customer_id','customer_unique_id']], on= 'customer_id', how = 'left')
```

```python
orders_done = orders_join[orders_join['order_status'] == 'delivered']
```

#### how long is long ? Lets say 90 days is the time period within which if a customer hasn't purchased. The customer is considered churned 

#### to decide what should be threshold for churn, we will use the full dataset 

```python
# Using full dataset to understand behavior

dfc = orders_done.sort_values(['customer_unique_id', 'order_purchase_timestamp'])

dfc['prev_date'] = dfc.groupby('customer_unique_id')['order_purchase_timestamp'].shift(1)
dfc['time_bw_orders'] = (dfc['order_purchase_timestamp'] - dfc['prev_date']).dt.days

time_bw_orders = dfc['time_bw_orders'].dropna()
```

```python
time_bw_orders.value_counts()
```

```python
# Exclude gaps <= 2 days: these represent orders placed within hours of each other
# (likely same-session split orders), not genuine repeat purchase behavior
time_bw_orders_filtered = dfc[(dfc['time_bw_orders'] >2) & (dfc['time_bw_orders'] <= 180) ]['time_bw_orders']
```

```python

plt.figure(figsize=(15,5))
plt.subplot(1,2,1)
plt.hist(time_bw_orders_filtered, bins=50)
plt.xlabel('Days between orders')
plt.ylabel('Frequency')
plt.title('Distribution of Inter-Order Time (>= 10 days)')
plt.subplot(1,2,2)
plt.hist(time_bw_orders_filtered, bins=50, log=True)
plt.xlabel('Days between orders')
plt.ylabel('Log Frequency')
plt.title('Distribution (Log Scale)')

plt.show()
```

```python
time_bw_orders_filtered.quantile([0.5,0.7,0.8,0.9,0.95])
```

```python
import numpy as np

sorted_gaps = np.sort(time_bw_orders_filtered)
cdf = np.arange(1, len(sorted_gaps)+1) / len(sorted_gaps)

plt.figure()
plt.plot(sorted_gaps, cdf)

plt.xlabel('Days between orders')
plt.ylabel('Cumulative probability')
plt.title('CDF of Inter-Order Time')
plt.axvline(x=90)  # threshold
plt.axvline(x=120, linestyle='--')
plt.show()
```

###### Its clear that ~ 75 % of repeat purchases happen within 90 days of the first order. 

```python
#
#purchase_date_max = dfc['order_purchase_timestamp'].max()
max_date_done = orders_done['order_purchase_timestamp'].max()
last_purchase_customer = orders_done.groupby('customer_unique_id')['order_purchase_timestamp'].max()

recent_day = (max_date_done - last_purchase_customer).dt.days

churn_90 = (recent_day > 90)
churn_120 = (recent_day > 120)
print(churn_90.mean(), churn_120.mean())
```

```python

changed = ((churn_90 == 1) & (churn_120 == 0)).sum()
total = len(recent_day)

changed_pct = changed / total
print(changed_pct)
```

While extending the churn threshold from 90 to 120 days reduces churn by ~7%, the majority of inactive customers are already captured within the 90-day window. This suggests that 90 days is sufficient to identify most disengaged users, with only a small proportion exhibiting delayed repeat behavior.

Customers whose last purchase occurred within the final 90 days of the dataset were excluded, as their churn status cannot be fully observed within the available time window.

```python
cutoff_date = max_date_done - pd.Timedelta(days=90)
```

```python
cutoff_date
```

```python
lpc = last_purchase_customer.reset_index()
valid_customers = lpc[
    lpc['order_purchase_timestamp'] <= cutoff_date
]['customer_unique_id']
```

```python
# Now create modeling dataset


orders_done_filtered = orders_done[
    orders_done['customer_unique_id'].isin(valid_customers)
]
```

```python
len(valid_customers)
```

```python
# last purchase per customer
last_purchase = orders_done_filtered.groupby('customer_unique_id')['order_purchase_timestamp'].max().reset_index()
max_date = orders_done_filtered['order_purchase_timestamp'].max()
last_purchase['date_since_purchase'] = (max_date - last_purchase['order_purchase_timestamp']).dt.days
last_purchase.head()
```

```python
last_purchase['churn']= last_purchase['date_since_purchase']>90
```

```python
last_purchase['churn'].value_counts()
```

```python
last_purchase['churn'].value_counts(normalize=True)*100
```

#### Here we see that about 73 % customers have churned. Most customers buy just once and business failed to retain them.

```python
# Total orders 
total_orders = orders_done_filtered.groupby('customer_unique_id')['order_id'].nunique().reset_index()
total_orders.rename(columns={'order_id': 'total_orders'}, inplace=True)
```

```python
# Total spend

order_items_sum = orders_items.groupby('order_id')['price'].sum().reset_index()
orders_price = pd.merge(orders_done_filtered, order_items_sum, on = 'order_id')
orders_spent = orders_price.groupby('customer_unique_id')['price'].sum().reset_index()
orders_spent.rename(columns={"price":"total_spent"},inplace = True)
```

```python
orders_spent.head()
```

```python
# Average order value
avg_order_value = orders_price.groupby('customer_unique_id')['price'].mean().reset_index()
avg_order_value.rename(columns = {'price':'avg_order_value'},inplace = True)
```

```python
# customer_df is built incrementally — run cells top-to-bottom in order

customer_df = last_purchase.merge(total_orders, on = 'customer_unique_id')
customer_df = customer_df.merge(orders_spent, on = 'customer_unique_id')
customer_df = customer_df.merge(avg_order_value, on = 'customer_unique_id')
```

```python
products = pd.read_csv("../data/olist_products_dataset.csv", dtype=str)
```

```python
orders_products = orders_done_filtered.merge(orders_items, on='order_id')
orders_products = orders_products.merge(products, on='product_id')
```

```python
def get_category(x):
    modes = x.mode()
    if len(modes) == 1:
        return modes[0]
    else:
        return 'multi'

customer_category = (
    orders_products
    .groupby('customer_unique_id')['product_category_name']
    .agg(get_category)
    .reset_index()
)
```

```python
customer_df = customer_df.merge(customer_category, on='customer_unique_id')
```

```python
# 
```

```python
orders_products.groupby('product_category_name')['order_id'].count().sort_values(ascending=False)
```

```python
#Churn by category 
customer_df.groupby('product_category_name')['churn'].mean().sort_values(ascending=False).reset_index()
```

```python
counts = customer_df['product_category_name'].value_counts()

valid_categories = counts[counts > 50].index  

df_plot = (
    customer_df[customer_df['product_category_name'].isin(valid_categories)]
    .groupby('product_category_name')['churn']
    .mean()
    .sort_values(ascending=False)
)
```

```python
import matplotlib.pyplot as plt

df_plot.plot(kind='barh', figsize=(10,6))

plt.xlabel('Churn Rate')
plt.ylabel('Product Category')
plt.title('Churn Rate by Product Category')

plt.gca().invert_yaxis()
plt.show()
```

```python
df_plot.head(15).plot(kind='barh', figsize=(8,5))

plt.title('Top 15 Categories by Churn')
plt.xlabel('Churn Rate')
plt.gca().invert_yaxis()
plt.show()
```

```python
customer_df['is_multi'] = (customer_df['product_category_name'] == 'multi').astype(int)
customer_df.groupby('is_multi')['churn'].mean()
```

```python
customer_df.groupby('is_multi')['churn'].value_counts()
```

#### Customers who purchase across multiple categories show lower churn compared to single-category buyers, suggesting that engagement breadth is a stronger indicator of retention than individual product category.

## Business interpretation


### Type 1 — Single-category buyers
#### Purchase with a specific intent, typically complete one transaction, and show high likelihood of churn. 

### Type 2 — Multi-category buyers
#### Exhibit broader browsing behavior, higher engagement with the platform, and a comparatively lower likelihood of churn.

```python
customer_df.tail()
```

```python
customer_df.groupby(['is_multi','is_late'])['churn'].mean()
```

Delivery delay does not increase churn in this dataset. Delivery cannot be the primary driver of churn

```python
import matplotlib.pyplot as plt

pivot = customer_df.groupby(['is_multi','is_late'])['churn'].mean().unstack()

pivot.plot(kind='bar', figsize=(8,5))

plt.ylabel('Churn Rate')
plt.xlabel('Multi-category Customer')
plt.title('Impact of Delivery Delay on Churn')
plt.xticks([0,1], ['Single-category','Multi-category'], rotation=0)
plt.legend(['Not Late','Late'])
plt.show()
```

```python
import seaborn as sns

sns.heatmap(pivot, annot=True, fmt=".2f", cmap='coolwarm')

plt.title('Churn Rate by Customer Type and Delivery Status')
plt.xlabel('is_late')
plt.ylabel('is_multi')
plt.show()
```

```python
delta = pivot[1] - pivot[0]  # late - not late

delta.plot(kind='bar', figsize=(6,4))

plt.title('Churn Rate Difference: Late vs On-Time Delivery (by Customer Type)')
plt.ylabel('Churn Increase')
plt.axhline(0, color='black', linewidth=0.8, linestyle='--')
plt.xticks([0,1], ['Single-category','Multi-category'], rotation=0)
plt.show()
```

## Late delivery reduces churn for single-category buyers (-0.14) — the opposite of what's expected. This shows delivery quality does not drive return behavior. For multi-category buyers the effect is negligible (+0.05).
