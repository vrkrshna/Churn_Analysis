# 01 Data Preparation 
## Loading CSV data, data type conversion

## Imports


```python
from sqlalchemy import create_engine
import pandas as pd
```

```python
## Load Data

orders = pd.read_csv("../data/olist_orders_dataset.csv")
customers = pd.read_csv("../data/olist_customers_dataset.csv")
order_items = pd.read_csv("../data/olist_order_items_dataset.csv")
products = pd.read_csv("../data/olist_products_dataset.csv")
```

## Data Type Conversion


```python
# Orders — datetime
date_cols = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in date_cols:
    orders[col] = pd.to_datetime(orders[col], errors='coerce')


# Customers — numeric
customers['customer_zip_code_prefix'] = pd.to_numeric(
    customers['customer_zip_code_prefix'], errors='coerce'
)


# Order items — numeric
num_cols = ['order_item_id', 'price', 'freight_value']

for col in num_cols:
    order_items[col] = pd.to_numeric(order_items[col], errors='coerce')


# Order items — datetime
order_items['shipping_limit_date'] = pd.to_datetime(
    order_items['shipping_limit_date'], errors='coerce'
)
```

## Push to PostgreSQL


```python
engine = create_engine("postgresql://evuser:evpassword@localhost:5432/ecommerce")
```

```python
orders.to_sql("orders", engine, if_exists="replace", index=False)
customers.to_sql("customers", engine, if_exists="replace", index=False)
order_items.to_sql("order_items", engine, if_exists="replace", index=False)
products.to_sql("products", engine, if_exists="replace", index=False)
```
