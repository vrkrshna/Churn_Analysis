## 03 — Analysis

### Objective
### Explore differences between churned and retained customers.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
customer_df = pd.read_csv("../data/customer_df.csv")
```

### Purchase Timing Analysis (avg_days_between_orders)

```python
# Exclude customers with avg gap < 2 days — these placed orders within hours of each other,
# likely a single shopping session split into multiple orders, not genuine repeat behavior
customer_df_filtered = customer_df[
    customer_df['avg_days_between_orders'] >= 2
]
```

```python
plt.figure(figsize=(7,5))
sns.violinplot(
    x='churn',
    y='avg_days_between_orders',
    data=customer_df_filtered,
    inner='quartile'
)
# Mean overlay
sns.pointplot(
    x='churn',
    y='avg_days_between_orders',
    data=customer_df_filtered,
    estimator='mean',
    color='red',
    markers='D',
    linestyles=''
)


plt.title('Distribution of Purchase Gaps by Churn')
plt.tight_layout()
plt.show()
```

```python
customer_df_filtered.groupby('churn')['avg_days_between_orders'].mean()
```

```python
customer_df_filtered.groupby('churn')['avg_days_between_orders'].quantile([0.25,0.5,0.75])
```

### Statistical Tests
#### Mann–Whitney U

```python
from scipy import stats

churned_avg = customer_df[customer_df["churn"] == 1]["avg_days_between_orders"]
retained_avg = customer_df[customer_df["churn"] == 0]["avg_days_between_orders"]

stat, p_value = stats.mannwhitneyu(churned_avg, retained_avg, alternative="two-sided")
print(churned_avg.mean() , retained_avg.mean())
print(churned_avg.median(), retained_avg.median())
print(f"Mann-Whitney U | U={stat:,.0f} | p={p_value:.4f}")
if p_value < 0.05:
    print("✅ Significant: average days between orders differs between churned and retained customers (p < 0.05)")
```

```python
# Effect size — rank-biserial correlation (r)
n1, n2 = len(churned_avg), len(retained_avg)
r = 1 - (2 * stat) / (n1 * n2)
print(f"Effect size (rank-biserial r): {r:.3f}")
# r < 0.1 = negligible, 0.1–0.3 = small, 0.3–0.5 = medium, >0.5 = large
```


Purchase timing shows a statistically significant difference between churned and retained customers, 
with a moderate effect size (r ≈ 0.35), indicating meaningful behavioral separation.


Most churned users are 2-order customers, while retention improves modestly with higher order counts. However, this effect is limited in strength.

In contrast, purchase timing shows a clear behavioral signal: retained customers exhibit longer and more variable inter-purchase intervals, while churned customers display shorter, burst-like purchasing patterns followed by disengagement.

### Product Engagement (is_multi)
Customers are classified based on whether they purchase from multiple product categories.

```python
pd.crosstab(customer_df['is_multi'], customer_df['churn'], normalize='index')
ct = pd.crosstab(customer_df['is_multi'], customer_df['churn'], normalize='index')
ax = ct.plot(kind='bar',width=0.8, figsize=(8,6))

# Labels
ax.set_xlabel('Customer Type')
ax.set_ylabel('Proportion')
ax.set_title('Churn Distribution by Category Engagement')

# X-ticks
ax.set_xticks([0, 1])
ax.set_xticklabels(['Single-category', 'Multi-category'], rotation=0)

# Legend
ax.legend(['Retained (0)', 'Churned (1)'])

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import chi2_contingency

ct = pd.crosstab(customer_df['is_multi'], customer_df['churn'])
chi2, p, dof, expected = chi2_contingency(ct)
print(f"Chi-square = {chi2:.2f} | p = {p:.4f} | dof = {dof}")
```

```python
n = ct.sum().sum()
phi2 = chi2 / n
r, k = ct.shape

cramers_v = np.sqrt(phi2 / min(k-1, r-1))

print(f"Cramér's V: {cramers_v:.4f}")
```

Although the relationship between category breadth (is_multi) and churn is statistically significant,
the effect size is very small (Cramér’s V ≈ 0.09), indicating limited practical impact.

### Order Count vs Churn


```python
customer_df.groupby('churn')['total_orders'].value_counts() 
```

```python
# Create crosstab
ct = pd.crosstab(customer_df['total_orders'], customer_df['churn'])

# Normalize (important)
ct_norm = ct.div(ct.sum(axis=0), axis=1)

# Plot
ct_norm.plot(kind='bar', figsize=(8,5))

plt.xlabel('Total Orders')
plt.ylabel('Proportion')
plt.title('Order Count Distribution by Churn')
plt.legend(title='Churn', labels=['Retained (0)', 'Churned (1)'])
plt.xticks( rotation = 0)
plt.tight_layout()
plt.show()
```

```python
from scipy.stats import chi2_contingency

# Build contingency table: counts of churned/not-churned by group

ct = pd.crosstab(customer_df['total_orders'], customer_df['churn'])

chi2, p, dof, expected = chi2_contingency(ct)
print(f"Chi-square = {chi2:.2f} | p = {p:.4f} | dof = {dof}")
```

```python

n = ct.sum().sum()
phi2 = chi2 / n
r, k = ct.shape

cramers_v = np.sqrt(phi2 / min(k-1, r-1))

print(f"Cramér's V: {cramers_v:.4f}")
```

There is a statistically significant relationship between order count and churn,
but the effect size is weak (Cramér’s V ≈ 0.09), indicating that order count alone
is not a strong predictor of churn.

###  Delivery Delay Analysis


```python
customer_df.groupby('churn')['avg_delivery_delay'].mean()
```

```python
customer_df.groupby('churn')['avg_delivery_delay'].quantile([0.25, 0.5, 0.75])
```

```python
plt.figure(figsize=(7,5))
sns.violinplot(
    x='churn',
    y='avg_delivery_delay',
    data=customer_df,
    inner='quartile'
)
# Mean overlay
sns.pointplot(
    x='churn',
    y='avg_delivery_delay',
    data=customer_df,
    estimator='mean',
    color='red',
    markers='D',
    linestyles=''
)
plt.axhline(0, color='black', linewidth=0.8, linestyle='--')


plt.title('Distribution of Average Delivery Delay by Churn')
plt.tight_layout()
plt.show()
```

Delivery delay does not appear to be a meaningful factor, as churned and retained customers exhibit nearly identical distributions and central tendencies.


## Overall Conclusion

Customer churn in this dataset is primarily driven by behavioral timing patterns rather than transactional metrics.

- Purchase timing (avg_days_between_orders) shows a clear and meaningful separation between churned and retained customers (moderate effect size).
- Category breadth (is_multi) and order count exhibit statistically significant but weak relationships with churn.
- Delivery delay shows no meaningful impact on churn behavior.

These findings suggest that churn is characterized by short, burst-like engagement followed by disengagement, rather than gradual decline or dissatisfaction with service factors.
